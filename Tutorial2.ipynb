```python
#1. Create a numpy array of size 2 x 2

import numpy as np

array_2x2 = np.array([
    [1,2],
    [3,4]
])

print('This is my numpy 2x2 array.')
print(array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4

array_3x4 = np.random.rand(3,4)

print('This is my random numpy array size 3 x 4:')
print(array_3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes

array_1 = np.ones((3,3))
array_0 = np.zeros((3,3))

print(array_1)
print(array_0)
```
